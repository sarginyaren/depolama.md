```python
import numpy as np
import pandas as pd
```

```python
satis1={"Tarih":["01.08.2026","01.08.2026","02.08.2026"],
        "Ürün":["Süt","Ekmek","Kahve"],
        "Kategori":["İçecek","Fırın","İçecek"],
        "Adet":[3,np.nan,2],
        "Birim Fiyat":[45,15,180]
        }
```

```python
dataFrame1=pd.DataFrame(satis1)
dataFrame1.to_excel("market_verisi.xlsx",index=False)
```

```python
dataFrame1['Toplam Tutar'] = dataFrame1['Adet'] * dataFrame1['Birim Fiyat']
dataFrame1
```

```python

en_cok_satilan_urun = dataFrame1.groupby("Ürün")["Adet"].sum().idxmax()
print(f"En fazla satılan ürün: {en_cok_satilan_urun}")
```

```python
en_cok_gelir_getiren_urun=dataFrame1.groupby("Ürün")["Toplam Tutar"].sum().idxmax()
print(f"En fazla gelir getiren ürün: {en_cok_gelir_getiren_urun}")
```
